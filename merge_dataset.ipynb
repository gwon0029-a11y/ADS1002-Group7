```python
import pandas as pd
from pathlib import Path
import openpyxl
```

# Demand Data


###### The following code merges all the data files in the folder Original Dataset > Demand Data > ALL_DEMAND_DATA into a single csv file 'Demand Data.csv' which is stored in the folder Merged Dataset.

```python
demand_df_list = [] #list of demand df: each element in the list contains data in a csv file
regions = ['NSW1', 'QLD1', 'SA1', 'VIC1', 'TAS1']
#iterate through all files in the demand data folder
for region in regions:
    for year in range (2000, 2020):
        for month in range (1, 13):
            #file_name e.g. 'Original Dataset/Demand Data/ALL_DEMAND_DATA/DATA200001_NSW1.csv'
            file_name = 'Original Dataset/Demand Data/ALL_DEMAND_DATA/DATA' + str(year) + str(month).zfill(2) + '_' + region + '.csv'
            if Path(file_name).is_file(): #if file exists
                df = pd.read_csv(file_name) #convert csv file to df
                demand_df_list.append(df) #append df to demand_df_list

demand_df = pd.concat(demand_df_list) #merge all df in demand_df_list
demand_df.to_csv('Merged Dataset/Demand Data.csv', index = False) #save merged df to a csv file
```
