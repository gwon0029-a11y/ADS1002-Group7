# Demand Data


###### The following code merges all the data files in the folder Original Dataset > Demand Data > ALL_DEMAND_DATA into a single csv file 'Demand Data.csv' which is stored in the folder Merged Dataset.

```python
demand_df_list = [] #list of demand df: each element in the list contains data in a csv file
regions = ['NSW1', 'QLD1', 'SA1', 'VIC1', 'TAS1']
#iterate through all files in the demand data folder
for region in regions:
    for year in range (2000, 2020):
        for month in range (1, 13):
            #file_name e.g. 'Original Dataset/Demand Data/ALL_DEMAND_DATA/DATA200001_NSW1.csv'
            file_name = 'Original Dataset/Demand Data/ALL_DEMAND_DATA/DATA' + str(year) + str(month).zfill(2) + '_' + region + '.csv'
            if Path(file_name).is_file(): #if file exists
                df = pd.read_csv(file_name) #convert csv file to df
                demand_df_list.append(df) #append df to demand_df_list

demand_df = pd.concat(demand_df_list) #merge all df in demand_df_list
demand_df.to_csv('Merged Dataset/Demand Data.csv', index = False) #save merged df to a csv file
```

```python
#reads a section from a text file and stores the lines in a list
def read_section(text_file, start_line, end_line, start_character):
    list = []
    with open(text_file, "r") as file:
        lines = file.read().splitlines()
        for i in range(start_line, end_line + 1):
            line = lines[i]
            column = line[start_character:]
            list.append(column)
    return list

temperature_columns = read_section(Path('Original Dataset/Temperature Data/HM01X_Notes_999999999743964.txt'), 19, 44, 29)
#station_columns = read_section(Path('Original Dataset/Temperature Data/HM01X_Notes_999999999743964.txt'), 398, 419, 29)
```

```python
#convert all temperature data txt files to csv files
for file in Path('Original Dataset/Temperature Data/').iterdir():
    if 'Data' in file.name: #iterate through every data file in Temperature Data folder
        with open('Original Dataset/Temperature Data/' + file.name, "r") as temperature_txt_file:
            df = pd.read_csv('Original Dataset/Temperature Data/' + file.name, sep = ",")
            df.to_csv('Merged Dataset/' + file.stem + '.csv', index = False)
```

```python
#read all temperature csv files
temperature_df_list = []
for file in Path('Merged Dataset/').iterdir():
    if 'HM01X_Data' in file.name: #iterate through every data file in Temperature Data folder
        with open('Merged Dataset/' + file.name, "r") as temperature_txt_file:
            file_name = 'Merged Dataset/' + file.name
            df = pd.read_csv(file_name)
            temperature_df_list.append(df)

temperature_df = pd.DataFrame(columns = temperature_columns)
temperature_df = pd.concat(temperature_df_list) #merge all df in demand_df_list
#temperature_df.to_csv('Merged Dataset/Merged Temperature Data.csv', index = False) #save merged df to a csv file
```
